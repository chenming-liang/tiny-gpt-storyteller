```python
import sys
sys.path.extend(["..", "../src"])

from datasets import load_dataset
from finetune import format_prompt

dataset = load_dataset("tatsu-lab/alpaca", split="train")
print("字段:", dataset[0].keys())
print("总条数:", len(dataset))
```

```python
# 用和 finetune 时相同的 tokenizer（自定义 4096 词表）
from pretrain import train_tokenizer

tokenizer = train_tokenizer(4096)
lengths = [len(tokenizer.encode(format_prompt(dataset[i]))) for i in range(2000)]

print(f"最大: {max(lengths)}")
print(f"最小: {min(lengths)}")
print(f"平均: {sum(lengths)/len(lengths):.0f}")
print(f"超过256的: {sum(1 for l in lengths if l > 256)}/{len(lengths)} ({sum(1 for l in lengths if l > 256)/len(lengths)*100:.0f}%)")
print(f"超过512的: {sum(1 for l in lengths if l > 512)}/{len(lengths)} ({sum(1 for l in lengths if l > 512)/len(lengths)*100:.0f}%)")
print(f"超过1024的: {sum(1 for l in lengths if l > 1024)}/{len(lengths)} ({sum(1 for l in lengths if l > 1024)/len(lengths)*100:.0f}%)")
```
