## 加载 Tinystory

```python
from datasets import load_dataset
import random

dataset = load_dataset('roneneldan/TinyStories')
samples = random.sample(list(dataset['train']), 5)
for i, s in enumerate(samples):
    print(f"--- Sample {i+1} ---")
    print(s["text"][:500])    # 太长就截断，只打前 500 字符
    print()
```

## Tokenizer

```python
from transformers import AutoTokenizer
tokenizer = AutoTokenizer.from_pretrained('gpt2')
tokenizer.pad_token = tokenizer.eos_token
print(tokenizer.pad_token)
tokens = tokenizer("I'm a student from tsinghua university.")
print(tokens['input_ids'])
print(tokenizer.decode(tokens['input_ids']))
tokens = tokenizer("I'm a student from Tsinghua university.")
print(tokens['input_ids'])
print(tokenizer.decode(tokens['input_ids']))
```

## Parameters

```python
sample = dataset["train"].select(range(10000))
lengths = [len(tokenizer.encode(item["text"])) for item in sample]

print(f"平均长度: {sum(lengths)/len(lengths):.0f}")
print(f"超过 256 的比例: {sum(1 for l in lengths if l > 256)/len(lengths)*100:.1f}%")
print(f"超过 512 的比例: {sum(1 for l in lengths if l > 512)/len(lengths)*100:.1f}%")
```
